```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 5))
df[['Age','Fare']].boxplot()
plt.title("Outlier Detection in Age and Fare")
plt.ylabel("Values")
plt.show()
```

```python
def detect_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_limit) | (df[column] > upper_limit)]
    print(f" {column} outliers:", len(outliers))
    detect_outliers(df, 'Age')
    detect_outliers(df, 'Fare')
```

```python
print("Age Outliers:", 
      ((df['Age'] < df['Age'].quantile(0.25) - 1.5 * (df['Age'].quantile(0.75) - df['Age'].quantile(0.25))) |
       (df['Age'] > df['Age'].quantile(0.75) + 1.5 * (df['Age'].quantile(0.75) - df['Age'].quantile(0.25)))).sum())

print("Fare Outliers:", 
      ((df['Fare'] < df['Fare'].quantile(0.25) - 1.5 * (df['Fare'].quantile(0.75) - df['Fare'].quantile(0.25))) |
       (df['Fare'] > df['Fare'].quantile(0.75) + 1.5 * (df['Fare'].quantile(0.75) - df['Fare'].quantile(0.25)))).sum())
```
